## Modelo a implementar
**Regresión Logística** para clasificación binaria

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
```

## 1. Carga del Dataset

Se descarga el dataset desde UCI ML Repository.
Los nombres de atributos son anonimizados por privacidad.
Se seleccionan solo las features numéricas.

```python
# Nombres de columnas
columns = ['A1', 'A2', 'A3', 'A4', 'A5', 'A6', 'A7', 'A8', 'A9', 'A10',
           'A11', 'A12', 'A13', 'A14', 'A15', 'A16']

# Cargar dataset desde archivo local
data = pd.read_csv('crx.data', names=columns, na_values='?')

# Convertir columnas numéricas
data['A2'] = pd.to_numeric(data['A2'], errors='coerce')
data['A14'] = pd.to_numeric(data['A14'], errors='coerce')

# Rellenar nulos con mediana
for col in ['A2', 'A3', 'A8', 'A11', 'A14', 'A15']:
    if data[col].isnull().sum() > 0:
        data[col] = data[col].fillna(data[col].median())

# Codificar variable objetivo
data['approved'] = (data['A16'] == '+').astype(int)

# Definir features y variable objetivo
FEATURES = ['A2', 'A3', 'A8', 'A11', 'A14', 'A15']

TARGET = 'approved'

# Extraer X e y
X = data[FEATURES].values.astype(np.float64)
y = data[TARGET].values.astype(np.float64)
m = y.size

print(f'Dimensiones dataset : {data.shape}')
print(f'X shape             : {X.shape}')
print(f'y shape             : {y.shape}')
print(f'm                   : {m:,}')
print(f'n features          : {len(FEATURES)}')
print(f'\nDistribución target:')
print(data[TARGET].value_counts())
```
