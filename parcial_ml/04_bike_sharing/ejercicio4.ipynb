

**Dataset:** Bike Sharing Demand  
**Fuente:** https://www.kaggle.com/c/bike-sharing-demand

## Variable objetivo (Y)
`count` — Número total de bicicletas alquiladas por hora

## Features seleccionadas (n=8)

| # | Feature | Descripción | Justificación |
|---|---|---|---|
| 1 | `season` | Estación del año (1-4) | En verano hay más demanda |
| 2 | `holiday` | Si es día festivo (0/1) | Los festivos cambian patrones |
| 3 | `workingday` | Si es día laboral (0/1) | Días laborales tienen picos diferentes |
| 4 | `weather` | Condición climática (1-4) | Mejor clima más alquileres |
| 5 | `temp` | Temperatura en Celsius | Temperaturas agradables aumentan demanda |
| 6 | `atemp` | Sensación térmica en Celsius | Lo que siente el usuario importa |
| 7 | `humidity` | Humedad relativa en % | Alta humedad reduce alquileres |
| 8 | `windspeed` | Velocidad del viento | Mucho viento reduce demanda |
| 9 | `hour` | Hora del día (0-23) | Hay picos en horas laborales |

## Columnas descartadas

| Columna | Razón |
|---|---|
| `datetime` | Texto — se usa para extraer hour |
| `casual` | Componente del target — filtrado de información |
| `registered` | Componente del target — filtrado de información |

## Tres modelos a implementar
1. **Regresión Lineal Múltiple** con Descenso por Gradiente
2. **Regresión Polinómica** con Descenso por Gradiente
3. **Ecuación Normal** — solución exacta sin iteraciones

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
```

## 1. Carga del Dataset

Se carga el dataset usando Pandas.
Se extrae la hora desde la columna datetime.
Se descartan casual y registered por ser componentes del target.

```python
# Cargar dataset
data = pd.read_csv('train.csv')

# Extraer hora desde datetime
data['datetime'] = pd.to_datetime(data['datetime'])
data['hour'] = data['datetime'].dt.hour

# Definir features y variable objetivo
FEATURES = [
    'season',
    'holiday',
    'workingday',
    'weather',
    'temp',
    'atemp',
    'humidity',
    'windspeed',
    'hour'
]

TARGET = 'count'

# Extraer X e y
X = data[FEATURES].values.astype(np.float64)
y = data[TARGET].values.astype(np.float64)
m = y.size

print(f'Dimensiones dataset : {data.shape}')
print(f'X shape             : {X.shape}')
print(f'y shape             : {y.shape}')
print(f'm                   : {m:,}')
print(f'n features          : {len(FEATURES)}')
```
