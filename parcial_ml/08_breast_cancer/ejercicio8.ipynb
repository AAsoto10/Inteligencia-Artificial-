# Regresión Logística
## Predicción de Cáncer de Mama - Breast Cancer Wisconsin

**Dataset:** Breast Cancer Wisconsin (Diagnostic)  
**Fuente:** https://archive.ics.uci.edu/ml/datasets/breast+cancer+wisconsin+(diagnostic)

## Variable objetivo (Y)
`diagnosis` — Diagnóstico del tumor (1=Maligno, 0=Benigno)

## Features seleccionadas (n=10)

| # | Feature | Descripción | Justificación |
|---|---|---|---|
| 1 | `radius_mean` | Radio promedio del tumor | Tamaño indica malignidad |
| 2 | `texture_mean` | Textura promedio | Textura irregular indica cáncer |
| 3 | `perimeter_mean` | Perímetro promedio | Relacionado con tamaño |
| 4 | `area_mean` | Área promedio | Tumores grandes son sospechosos |
| 5 | `smoothness_mean` | Suavidad promedio | Irregularidad indica cáncer |
| 6 | `compactness_mean` | Compacidad promedio | Forma del tumor |
| 7 | `concavity_mean` | Concavidad promedio | Contornos cóncavos |
| 8 | `concave_points_mean` | Puntos cóncavos promedio | Número de indentaciones |
| 9 | `symmetry_mean` | Simetría promedio | Asimetría indica malignidad |
| 10 | `fractal_dimension_mean` | Dimensión fractal promedio | Complejidad del borde |

## Columnas descartadas

| Columna | Razón |
|---|---|
| `id` | Identificador — sin relación con diagnóstico |
| `*_se` (10 cols) | Error estándar — redundante con mean |
| `*_worst` (10 cols) | Peor valor — redundante con mean |

## Modelo a implementar
**Regresión Logística** para clasificación binaria

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
```

## 1. Carga del Dataset

Se descarga el dataset desde UCI ML Repository.
Se seleccionan las 10 features de promedios (mean).
Se codifica el diagnóstico como binario.

```python
# Nombres de columnas
columns = ['id', 'diagnosis',
           'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean',
           'smoothness_mean', 'compactness_mean', 'concavity_mean',
           'concave_points_mean', 'symmetry_mean', 'fractal_dimension_mean',
           'radius_se', 'texture_se', 'perimeter_se', 'area_se',
           'smoothness_se', 'compactness_se', 'concavity_se',
           'concave_points_se', 'symmetry_se', 'fractal_dimension_se',
           'radius_worst', 'texture_worst', 'perimeter_worst', 'area_worst',
           'smoothness_worst', 'compactness_worst', 'concavity_worst',
           'concave_points_worst', 'symmetry_worst', 'fractal_dimension_worst']

# Cargar dataset desde UCI
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/breast-cancer-wisconsin/wdbc.data'
data = pd.read_csv(url, names=columns)

# Codificar variable objetivo (M=Maligno=1, B=Benigno=0)
data['diagnosis_encoded'] = (data['diagnosis'] == 'M').astype(int)

# Definir features (solo promedios) y variable objetivo
FEATURES = [
    'radius_mean', 'texture_mean', 'perimeter_mean', 'area_mean',
    'smoothness_mean', 'compactness_mean', 'concavity_mean',
    'concave_points_mean', 'symmetry_mean', 'fractal_dimension_mean'
]

TARGET = 'diagnosis_encoded'

# Extraer X e y
X = data[FEATURES].values.astype(np.float64)
y = data[TARGET].values.astype(np.float64)
m = y.size

print(f'Dimensiones dataset : {data.shape}')
print(f'X shape             : {X.shape}')
print(f'y shape             : {y.shape}')
print(f'm                   : {m:,}')
print(f'n features          : {len(FEATURES)}')
print(f'\nDistribución target:')
print(data[TARGET].value_counts())
```
