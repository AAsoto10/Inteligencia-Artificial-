# Regresión Logística
## Predicción de Ingresos - Adult Census

**Dataset:** Adult Census Income  
**Fuente:** https://archive.ics.uci.edu/ml/datasets/adult

## Variable objetivo (Y)
`income` — Si el ingreso es mayor a 50K (1=Sí, 0=No)

## Features seleccionadas (n=6)

| # | Feature | Descripción | Justificación |
|---|---|---|---|
| 1 | `age` | Edad del individuo | Mayor edad puede significar más experiencia |
| 2 | `education_num` | Años de educación | Más educación mayor ingreso |
| 3 | `hours_per_week` | Horas trabajadas por semana | Más horas más ingreso potencial |
| 4 | `capital_gain` | Ganancias de capital | Indica inversiones y riqueza |
| 5 | `capital_loss` | Pérdidas de capital | Indica actividad financiera |
| 6 | `fnlwgt` | Peso final del censo | Representatividad estadística |

## Columnas descartadas

| Columna | Razón |
|---|---|
| `workclass` | Categórica — requiere encoding |
| `education` | Redundante con education_num |
| `marital_status` | Categórica — requiere encoding |
| `occupation` | Categórica — requiere encoding |
| `relationship` | Categórica — requiere encoding |
| `race` | Categórica — requiere encoding |
| `sex` | Categórica — requiere encoding |
| `native_country` | Categórica — muchas categorías |

## Modelo a implementar
**Regresión Logística** para clasificación binaria

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
```

## 1. Carga del Dataset

Se descarga el dataset desde UCI ML Repository.
Se seleccionan solo las features numéricas.
Se codifica la variable objetivo como binaria.

```python
# Nombres de columnas del dataset
columns = [
    'age', 'workclass', 'fnlwgt', 'education', 'education_num',
    'marital_status', 'occupation', 'relationship', 'race', 'sex',
    'capital_gain', 'capital_loss', 'hours_per_week', 'native_country', 'income'
]

# Cargar dataset desde archivo local
data = pd.read_csv('adult.data', names=columns, sep=r',\s*', engine='python', na_values='?')

# Las features numéricas no tienen nulos, solo las categóricas
# No eliminamos filas, usamos solo features numéricas

# Codificar variable objetivo
data['income_encoded'] = (data['income'] == '>50K').astype(int)

# Definir features y variable objetivo
FEATURES = [
    'age',
    'education_num',
    'hours_per_week',
    'capital_gain',
    'capital_loss',
    'fnlwgt'
]

TARGET = 'income_encoded'

# Extraer X e y
X = data[FEATURES].values.astype(np.float64)
y = data[TARGET].values.astype(np.float64)
m = y.size

print(f'Dimensiones dataset : {data.shape}')
print(f'X shape             : {X.shape}')
print(f'y shape             : {y.shape}')
print(f'm                   : {m:,}')
print(f'n features          : {len(FEATURES)}')
print(f'\nDistribución target:')
print(data[TARGET].value_counts())
```
