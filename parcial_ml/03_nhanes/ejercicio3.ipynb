# Regresión Lineal Múltiple, Polinómica y Ecuación Normal
## Predicción de Colesterol Total - NHANES Laboratory

**Dataset:** NHANES (National Health and Nutrition Examination Survey) - Laboratory Data  
**Fuente:** https://www.cdc.gov/nchs/nhanes/

## Variable objetivo (Y)
`LBXTC` — Colesterol total en mg/dL

## Features seleccionadas (n=10)

| # | Feature | Descripción | Justificación |
|---|---|---|---|
| 1 | `LBDHDD` | HDL Colesterol (mg/dL) | Componente del colesterol total |
| 2 | `LBXHGB` | Hemoglobina (g/dL) | Indicador de salud general |
| 3 | `LBXGH` | Hemoglobina glicosilada (%) | Relacionado con metabolismo |
| 4 | `LBXHSCRP` | Proteína C reactiva (mg/L) | Marcador de inflamación |
| 5 | `LBXWBCSI` | Glóbulos blancos (1000 cells/uL) | Indicador inmunológico |
| 6 | `LBXRBCSI` | Glóbulos rojos (million cells/uL) | Salud sanguínea |
| 7 | `LBXPLTSI` | Plaquetas (1000 cells/uL) | Función de coagulación |
| 8 | `LBXHCT` | Hematocrito (%) | Porcentaje de glóbulos rojos |
| 9 | `LBXRDW` | Ancho de distribución de glóbulos rojos (%) | Variabilidad celular |
| 10 | `LBXMCVSI` | Volumen corpuscular medio (fL) | Tamaño promedio de glóbulos rojos |

## Columnas descartadas

| Columna | Razón |
|---|---|
| `SEQN` | Identificador — sin relación con colesterol |
| Columnas con >30% nulos | Datos insuficientes |
| Columnas de peso/muestreo | Variables administrativas |

## Tres modelos a implementar
1. **Regresión Lineal Múltiple** con Descenso por Gradiente
2. **Regresión Polinómica** con Descenso por Gradiente
3. **Ecuación Normal** — solución exacta sin iteraciones

```python
import numpy as np
import pandas as pd
from matplotlib import pyplot
```

## 1. Carga del Dataset

Se carga el dataset de laboratorio combinado.
Se seleccionan marcadores sanguíneos con buena correlación.
Se rellenan nulos con la mediana.

```python
# Cargar dataset
data = pd.read_csv('combined_laboratory.csv')

# Definir features (marcadores de laboratorio)
FEATURES = [
    'LBDHDD',     # HDL Colesterol
    'LBXHGB',     # Hemoglobina
    'LBXGH',      # Hemoglobina glicosilada
    'LBXHSCRP',   # Proteína C reactiva
    'LBXWBCSI',   # Glóbulos blancos
    'LBXRBCSI',   # Glóbulos rojos
    'LBXPLTSI',   # Plaquetas
    'LBXHCT',     # Hematocrito
    'LBXRDW',     # Ancho distribución glóbulos rojos
    'LBXMCVSI'    # Volumen corpuscular medio
]

TARGET = 'LBXTC'  # Colesterol total

# Rellenar valores nulos con la mediana
for col in FEATURES + [TARGET]:
    if data[col].isnull().sum() > 0:
        data[col] = data[col].fillna(data[col].median())

# Extraer X e y
X = data[FEATURES].values.astype(np.float64)
y = data[TARGET].values.astype(np.float64)
m = y.size

print(f'Dimensiones dataset : {data.shape}')
print(f'X shape             : {X.shape}')
print(f'y shape             : {y.shape}')
print(f'm                   : {m:,}')
print(f'n features          : {len(FEATURES)}')
```
