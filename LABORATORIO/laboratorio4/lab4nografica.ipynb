## ¿Qué es este dataset?

El Pump Sensor Data es un dataset de series temporales generado por sensores instalados en una bomba de agua industrial real. Captura lecturas de **52 sensores** que monitorean el comportamiento de la bomba durante operación normal y durante fallas. Contiene más de 220,000 registros con el estado operativo de la máquina como variable objetivo.

## 4. Normalización de Características

Las lecturas de los sensores tienen escalas muy diferentes.
Se aplica normalización Z-score antes de dividir los datos:

$$x_{norm} = \frac{x - \mu}{\sigma}$$

```python
def featureNormalize(X):
    X_norm = X.copy()
    mu    = np.mean(X, axis=0)
    sigma = np.std(X, axis=0)
    sigma[sigma == 0] = 1   # evitar división por cero
    X_norm = (X - mu) / sigma
    return X_norm, mu, sigma

X_norm, mu, sigma = featureNormalize(X)

print('Media calculada (≈ 0):', np.mean(X_norm, axis=0).round(3)[:5], '...')
print('Desv. estándar (≈ 1) :', np.std(X_norm, axis=0).round(3)[:5], '...')
```

## 7. Función Sigmoidea

$$g(z) = \frac{1}{1 + e^{-z}}$$

```python
def calcularSigmoide(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))

# Prueba
z = [-100, 0, 100]
print(f'g({z}) = {calcularSigmoide(z)}')
print(f'sigmoid(0) = {calcularSigmoide(0)} ← debe ser 0.5')
```

## 8. Función de Costo Regularizada — devuelve J y grad juntos

Igual que en los cuadernillos del inge (`lrCostFunction`).
Devuelve **J y grad en la misma función** para que `optimize.minimize` con `jac=True` funcione.

$$J(\theta) = \frac{1}{m}\sum_{i=1}^{m}\left[-y^{(i)}\log(h) - (1-y^{(i)})\log(1-h)\right] + \frac{\lambda}{2m}\sum_{j=1}^{n}\theta_j^2$$

**Nota:** No se regulariza $\theta_0$ (el término de bias).

```python
def calcularCosto(theta, X, y, lambda_):
    m = y.size

    # convierte las etiquetas a valores enteros si son boleanos
    if y.dtype == bool:
        y = y.astype(int)

    J    = 0
    grad = np.zeros(theta.shape)

    h = calcularSigmoide(X.dot(theta.T))
    h = np.clip(h, 1e-10, 1 - 1e-10)

    # No regularizar theta[0] (bias)
    temp    = theta.copy()
    temp[0] = 0

    J    = (1/m) * np.sum(-y.dot(np.log(h)) - (1-y).dot(np.log(1-h))) + (lambda_/(2*m)) * np.sum(np.square(temp))
    grad = (1/m) * (h - y).dot(X)
    grad = grad + (lambda_/m) * temp   # regularización en el gradiente

    return J, grad

# Prueba con valores del inge
theta_t = np.array([-2, -1, 1, 2], dtype=float)
X_t = np.concatenate((np.ones((5, 1)), np.arange(1, 16).reshape(5, 3, order='F')/10.0), axis=1)
y_t = np.array([1, 0, 1, 0, 1])
J_t, grad_t = calcularCosto(theta_t, X_t, y_t, 3)
print(f'Costo         : {J_t:.6f}')
print(f'Costo esperado: 2.534819')
print(f'Gradientes    : {grad_t}')
print(f'Grad esperados: [0.146561, -0.548558, 0.724722, 1.398003]')
```

## 9. Clasificación One-vs-All

Se entrena un clasificador logístico regularizado por cada clase
usando `scipy.optimize.minimize` con método BFGS, igual que en los cuadernillos del inge.

- Se pasa `jac=True` porque `calcularCosto` devuelve J y grad juntos
- Para cada clase c se crea un `y` binario: `(y == c)` → 1 si es clase c, 0 si no

```python
def oneVsAll(X, y, num_labels, lambda_):
    m, n = X.shape
    all_theta = np.zeros((num_labels, n + 1))

    # Agrega unos a la matriz X
    X = np.concatenate([np.ones((m, 1)), X], axis=1)

    for c in np.arange(num_labels):
        initial_theta = np.zeros(n + 1)
        options = {'maxiter': 100}

        res = optimize.minimize(calcularCosto,
                                initial_theta,
                                (X, (y == c), lambda_),   # y binario: ¿es clase c?
                                jac=True,                 # calcularCosto devuelve J y grad
                                method='BFGS',
                                options=options)

        all_theta[c] = res.x
        print(f'  Clase {c}: J={res.fun:.4f}')

    return all_theta

print('Función oneVsAll definida ✓')
```

## 10. Entrenamiento del Modelo

Se llama a `oneVsAll` con los datos de entrenamiento.
El resultado `all_theta` tiene forma (3, n+1): una fila de parámetros por clase.

```python
lambda_   = 0.1

# oneVsAll agrega bias internamente → pasarle X SIN bias
# X_train tiene bias (columna de unos) → quitársela antes
X_train_nb = X_train[:, 1:]  # shape (m, 52)
X_val_nb   = X_val[:, 1:]    # shape (m, 52)

print(f'Entrenando {num_labels} clasificadores con lambda={lambda_}...')
print(f'Datos: {X_train_nb.shape[0]:,} ejemplos × {X_train_nb.shape[1]} features')
print()

all_theta = oneVsAll(X_train_nb, y_train, num_labels, lambda_)

print(f'\nall_theta shape: {all_theta.shape}  →  (num_labels, n+1)')
print('Entrenamiento completado ✓')
```

## 11. Gráfica de Costo — Convergencia por Clase

Se muestra el costo final alcanzado por cada clasificador
para verificar que todos convergieron correctamente.

```python
# Calcular costo final de cada clasificador
# all_theta fue entrenado con X que tiene bias → espera (n+1) = 53 columnas
# Agregamos bias a X_train_nb para que cuadre
costos_finales = []
nombres_clases = ['NORMAL', 'BROKEN', 'RECOVERING']
colores = ['#4ade80', '#f87171', '#fbbf24']

X_train_bias = np.concatenate([np.ones((X_train_nb.shape[0], 1)), X_train_nb], axis=1)

for c in range(num_labels):
    y_bin = (y_train == c).astype(float)
    J_c, _ = calcularCosto(all_theta[c], X_train_bias, y_bin, lambda_)
    costos_finales.append(J_c)

pyplot.figure(figsize=(8, 4))
bars = pyplot.bar(nombres_clases, costos_finales, color=colores)
for bar, val in zip(bars, costos_finales):
    pyplot.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.002,
                f'{val:.4f}', ha='center', fontsize=10, fontweight='bold')
pyplot.title('Costo Final por Clasificador — One-vs-All')
pyplot.xlabel('Clase')
pyplot.ylabel('Costo J')
pyplot.grid(axis='y', alpha=0.4)
pyplot.tight_layout()
pyplot.show()

print('Costos finales:')
for c, (nombre, costo) in enumerate(zip(nombres_clases, costos_finales)):
    print(f'  Clase {c} ({nombre}): J = {costo:.4f}')
```

## 12. Predicción One-vs-All

Se calcula la probabilidad de cada clase y se asigna la de mayor probabilidad:

$$\hat{y} = \arg\max_c \; sigmoid(\theta_c^T x)$$

Igual que `predictOneVsAll` del inge.

```python
def predictOneVsAll(all_theta, X):
    m = X.shape[0]
    num_labels = all_theta.shape[0]

    p = np.zeros(m)

    # Agrega unos a la matriz X
    X = np.concatenate([np.ones((m, 1)), X], axis=1)
    p = np.argmax(calcularSigmoide(X.dot(all_theta.T)), axis=1)

    return p

print('Función predictOneVsAll definida ✓')
```

## 13. Evaluación del Modelo — Precisión en Entrenamiento y Validación

Se evalúa la precisión primero en entrenamiento y luego en validación
(datos que el modelo **nunca vio**).

```python
# predictOneVsAll agrega bias internamente → pasarle X SIN bias
pred_train = predictOneVsAll(all_theta, X_train_nb)
pred_val   = predictOneVsAll(all_theta, X_val_nb)

prec_train = np.mean(pred_train == y_train) * 100
prec_val   = np.mean(pred_val   == y_val)   * 100

print('=' * 50)
print('       EVALUACIÓN DEL MODELO')
print('=' * 50)
print(f'Precisión entrenamiento : {prec_train:.2f}%')
print(f'Precisión validación    : {prec_val:.2f}%')
print('=' * 50)
print('\nPrecisión por clase (validación):')
for c in range(num_labels):
    mask = y_val == c
    if mask.sum() > 0:
        acc = np.mean(pred_val[mask] == c) * 100
        print(f'  Clase {c} ({nombres_clases[c]}): {acc:.2f}%  ({mask.sum()} ejemplos)')
```

## 14. Gráfica de Precisión por Clase

Se visualiza la efectividad del modelo en cada clase de la validación.

```python
prec_por_clase = []
for c in range(num_labels):
    mask = y_val == c
    acc  = np.mean(pred_val[mask] == c) * 100 if mask.sum() > 0 else 0
    prec_por_clase.append(acc)

pyplot.figure(figsize=(8, 4))
bars = pyplot.bar(nombres_clases, prec_por_clase, color=colores)
pyplot.axhline(y=prec_val, color='gray', linestyle='--',
               label=f'Precisión promedio: {prec_val:.1f}%')
for bar, v in zip(bars, prec_por_clase):
    pyplot.text(bar.get_x() + bar.get_width()/2, v + 0.5,
                f'{v:.1f}%', ha='center', fontsize=10, fontweight='bold')
pyplot.title('Precisión por Clase — Validación')
pyplot.xlabel('Clase')
pyplot.ylabel('Precisión (%)')
pyplot.ylim(0, 110)
pyplot.legend()
pyplot.grid(axis='y', alpha=0.4)
pyplot.tight_layout()
pyplot.show()
```

## 15. Predicciones sobre Datos de Validación

Se muestran las predicciones del modelo sobre un subconjunto de validación,
comparando el valor real con la predicción, igual que en los cuadernillos del inge.

```python
print(X_val_nb.shape)
print('Precision del conjunto de validacion: {:.2f}%'.format(prec_val))

# Tomar 30 ejemplos de validación — X_val_nb sin bias, predictOneVsAll lo agrega
XPrueba = X_val_nb[:30, :].copy()
print(XPrueba.shape)

XPrueba_con_bias = np.concatenate([np.ones((30, 1)), XPrueba], axis=1)
print(XPrueba_con_bias.shape)

p = np.argmax(calcularSigmoide(XPrueba_con_bias.dot(all_theta.T)), axis=1)
print('\nPredicciones:', p)
print('Reales      :', y_val[:30].astype(int))
print('\nDetalle:')
print(f'{"#":<4} {"Real":<12} {"Predicho":<12} {"OK":>4}')
print('-' * 35)
for i in range(30):
    r  = int(y_val[i])
    pr = int(p[i])
    ok = '✓' if r == pr else '✗'
    print(f'{i+1:<4} {nombres_clases[r]:<12} {nombres_clases[pr]:<12} {ok:>4}')
```
